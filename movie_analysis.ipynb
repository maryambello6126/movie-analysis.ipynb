```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
df.describe()
```

```python
avg_tip_by_day = df.groupby("day")["tip"].mean().sort_values(ascending=False)
avg_tip_by_day.plot(kind="bar", title="Average Tip by Day")
plt.show()
```

```python
df.groupby("time")["tip"].mean().plot(kind="bar", title="Average Tip by Time")
plt.show()
```
